````python
# Ejercicio 1
from tkinter import messagebox
from tkinter import *
class Contactos:
    def __init__(self, master):
        self.master = master

        self.frame_formulario = Frame(self.master)

        self.label_nombre = Label(self.frame_formulario, text='Nombre: ')
        self.label_nombre.grid(padx=5, pady=5, row=0, column=0)
        self.entry_nombre = Entry(self.frame_formulario)
        self.entry_nombre.grid(padx=5, pady=5, row=0, column=1)

        self.label_apellidos = Label(self.frame_formulario, text='Apellidos: ')
        self.label_apellidos.grid(padx=5, pady=5, row=1, column=0)
        self.entry_apellidos = Entry(self.frame_formulario)
        self.entry_apellidos.grid(padx=5, pady=5, row=1, column=1)

        self.label_telefono = Label(self.frame_formulario, text='Teléfono: ')
        self.label_telefono.grid(padx=5, pady=5, row=2, column=0)
        self.entry_telefono = Entry(self.frame_formulario)
        self.entry_telefono.grid(padx=5, pady=5, row=2, column=1)

        self.frame_formulario.pack()

        self.frame = Frame(self.master)
        self.boton_agregar = Button(self.frame, text = 'Agregar Contacto', command = lambda:
                                    self.agregar_contacto(self.entry_nombre.get(),
                                                          self.entry_apellidos.get(),
                                                          self.entry_telefono.get()))
        self.boton_agregar.pack(padx=5, pady=5, side="left")

        self.boton_eliminar = Button(self.frame, text='Eliminar Contacto', command = self.eliminar_contacto)
        self.boton_eliminar.pack(padx=5, pady=5, side="left")


        self.lista_contactos = []
        self.list_var = StringVar(value=[])

        list_variable = Variable(value=self.lista_contactos)
        self.listbox_contacto = Listbox(self.frame, listvariable=list_variable)
        self.listbox_contacto.pack(padx=5, pady=10)

        self.frame.pack()


    def agregar_contacto(self, nombre, apellidos, telefono):
        if nombre == '' or apellidos == '' or telefono == '':
            messagebox.showerror("Error", "Todos los campos son obligatorios.")
            return
        contacto = f"{nombre} {apellidos} - {telefono}"
        self.lista_contactos.append(contacto)
        self.actualizar_lista_contactos()
        messagebox.showinfo("Contacto Agregado", f"Contacto {nombre} {apellidos} agregado exitosamente.")
        self.entry_nombre.delete(0, END)
        self.entry_apellidos.delete(0, END)
        self.entry_telefono.delete(0, END)

    def actualizar_lista_contactos(self):
        # Borrar todo
        self.listbox_contacto.delete(0, END)
        # Insertar de nuevo
        for contacto in self.lista_contactos:
            self.listbox_contacto.insert(END, contacto)

    def eliminar_contacto(self):
        seleccion = self.listbox_contacto.curselection()
        if not seleccion:
            messagebox.showerror("Error", "Seleccione un contacto para eliminar.")
            return
        indice = seleccion[0]
        contacto_eliminado = self.lista_contactos.pop(indice)
        self.actualizar_lista_contactos()
        messagebox.showinfo("Contacto Eliminado", f"Contacto {contacto_eliminado} eliminado exitosamente.")
        

root = Tk()
ventana = Contactos(root)
root.mainloop()
````

Crea una aplicación que muestre la hora actual (horas, minutos y segundos), actualizándose constantemente. Podemos utilizar una clase o no. Tengamos en cuenta lo siguiente:

- **Importaciones:** `import time`: Importa el módulo `time`, que nos permite trabajar con funciones relacionadas con el tiempo. En este caso, se usa para obtener la hora actual del sistema.
- **Creación de la etiqueta (`Label`) que muestra la hora:** Modifica la fuente, el tamaño y el estilo (Ej. `Label(root, font=('times', 30, 'bold'), bg='yellow')`)

- **Obtener la hora actual y actualizar la etiqueta:** Emplea `time.strftime('%H:%M:%S')` para obtener la hora actual del sistema y formatearla en el formato `HH:MM:SS` (hora, minutos y segundos). Compara después la hora actual que has obtenido con la hora anterior que ya tenías). Si la hora ha cambiado, significa que el reloj necesita actualizarse con `clock.config(text=nueva_hora)`. Todo esto lo puedes hacer dentro de una función.
- **Llamada repetida a la función anterior:** Emplea `mi_etiqueta.after(200, mi_función) para que la función se ejecute cada 200 milisegundos (0.2 segundos). Es una forma de crear un bucle de actualización. La función `after` permite crear una llamada asincrónica para que no se congele la interfaz mientras se actualiza el reloj.

````python
from tkinter import *
class TimerApp():
    def __init__(self, master):
        self.master = master
        self.master.title('Temorizador')
        self.master.geometry('300x200')

        self.frame = Frame(self.master)

        self.tiempo_restante = IntVar()
        self.tiempo_restante.set(10)

        self.running = False

        self.label_tiempo_restante = Label(self.frame, text=f'Tiempo restante: {self.tiempo_restante.get()}', font=('arial', 24))
        self.label_tiempo_restante.pack(padx=20, pady=20)

        self.boton = Button(self.frame, text='Iniciar/Detener', command=self.toggle_timer)
        self.boton.pack(padx=10, pady=10)


        self.frame.pack()
    
    def toggle_timer(self):
        if not self.running:
            self.running = True
            self.countdown()
        else:
            self.running = False

    def countdown(self):
        if self.running and self.tiempo_restante.get() > 0:
            self.tiempo_restante.set(self.tiempo_restante.get() - 1)
            self.label_tiempo_restante.config(text=f'Tiempo restante: {self.tiempo_restante.get()}')
            self.master.after(1000, self.countdown)
        elif self.tiempo_restante.get() == 0:
            self.running = False
            self.tiempo_restante.set(10)
            self.label_tiempo_restante.config(text='Tiempo Agotado!')


root = Tk()
timer_app = TimerApp(root)
root.mainloop()
````
